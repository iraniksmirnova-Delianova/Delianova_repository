Завдання 2. Аналіз повноти відповідей респондентів
Мета: порахувати кількість респондентів, у яких немає пропусків у всіх питаннях опитування.

1) Отримуємо список питань з survey_results_schema (поле qname)


```python
qnames = schema["qname"].dropna().unique()
print("Кількість qname у schema:", len(qnames))
print("Перші 10 qname:", qnames[:10])
```

2. Інтерсекція множин (фільтруємо колонки). Беремо тільки ті qname, які реально є в public.

```python
public_cols = set(public.columns)
schema_cols = set(qnames)

common_cols = sorted(public_cols.intersection(schema_cols))

print("Кількість колонок у public:", len(public.columns))
print("Кількість qname у schema:", len(qnames))
print("Кількість спільних колонок (public ∩ schema):", len(common_cols))
print("Перші 10 спільних колонок:", common_cols[:10])
```

3. Підрахунок респондентів без пропусків

```python
# Беремо тільки потрібні колонки
public_common = public[common_cols]

# Маска: True, якщо в рядку немає пропусків у жодній з цих колонок
no_missing_mask = public_common.notna().all(axis=1)

# Підрахунок респондентів, які відповіли на всі запитання
respondents_all_answered = int(no_missing_mask.sum())

print(respondents_all_answered)
```

Перевірка 1: скільки колонок реально перевіряємо

```python
print("К-сть qname у schema:", schema["qname"].nunique())
print("К-сть колонок у public:", public.shape[1])
print("К-сть спільних колонок (public ∩ schema):", len(common_cols))
```

Перевірка 2: максимум скільки питань взагалі заповнив "найповніший" респондент

```python
filled_per_row = public[common_cols].notna().sum(axis=1)

print("Максимум заповнених відповідей у одного респондента:", filled_per_row.max())
print("К-сть питань, які ми перевіряємо:", len(common_cols))
```

Так як filled_per_row.max() менше , ніж len(common_cols) - тоді 0 гарантовано і це підтвердження

Завдання 3. Статистичний аналіз досвіду респондентів.
Мета: обчислити міри центральної тенденції для змінної WorkExp
(досвід роботи респондентів у роках).
Показники:
- середнє (mean)
- медіана (median)
- мода (mode)


```python
#Перевіряємо що колонка існує
print("WorkExp" in public.columns)
# Дивимось тип і кількість пропусків
public["WorkExp"].info()
```

Очищення. Ігноруємо пропуски.

```python
work_exp = public["WorkExp"].dropna()

print("Кількість значень без пропусків:", work_exp.shape[0])
# Пропущені значення були виключені з аналізу, оскільки вони не містять числової інформації і можуть спотворювати результати обчислення мір центральної тенденції.
```

Обчислення статистик

```python
mean_exp = work_exp.mean()
median_exp = work_exp.median()
mode_exp = work_exp.mode().iloc[0]  # беремо першу моду, якщо їх кілька

stats = pd.DataFrame({
    "Показник": ["Середнє", "Медіана", "Мода"],
    "Значення": [mean_exp, median_exp, mode_exp]
})

stats
```

Виснновок.
Середній досвід роботи респондентів становить 13.37 року, тоді як медіана і мода дорівнюють 10 рокам.
Це означає, що більшість учасників опитування мають близько 10 років досвіду, а вищий середній показник зумовлений наявністю респондентів із дуже великим стажем, які “тягнуть” середнє значення вгору.
Отже, розподіл досвіду є асиметричним вправо, з невеликою кількістю дуже досвідчених спеціалістів.

Завдання 7. Географічний аналіз компенсації Python-розробників
Мета: серед респондентів, які програмують на Python, проаналізувати річну компенсацію
(ConvertedCompYearly) в розрізі країн. Для кожної країни обчислюються
середнє та медіанне значення компенсації.

```python
# Перевірка потрібних колонок
cols = ["LanguageHaveWorkedWith", "ConvertedCompYearly", "Country"]
print(all(col in public.columns for col in cols))

public[cols].head()
```

```python
# Фільтр Python + без пропусків у зарплаті
python_df = public[
    public["LanguageHaveWorkedWith"].str.contains("Python", na=False)
]

python_df = python_df.dropna(subset=["ConvertedCompYearly", "Country"])

print("Кількість Python-розробників з відомою компенсацією:", python_df.shape[0])
```

```python
# Групування по країнах
country_stats = (
    python_df
    .groupby("Country")["ConvertedCompYearly"]
    .agg(["mean", "median"])
    .reset_index()
)

country_stats.columns = ["Country", "MeanCompensation", "MedianCompensation"]

country_stats.head()
```

```python
# Відсортувати за середньою компенсацією
country_stats.sort_values("MeanCompensation", ascending=False).head(10)
```

Висновок
Серед Python-розробників найбільші середні та медіанні рівні компенсації спостерігаються в країнах з більш прогресивними ІТ-ринками та високою вартістю життя.
У топі за середньою компенсацією знаходяться Oman, Andorra, Viet Nam, United States та Switzerland.
Водночас різниця між середнім та медіанним значеннями у багатьох країнах (наприклад, United States, Singapore, Israel) свідчить про нерівномірний розподіл доходів: невелика кількість дуже високооплачуваних спеціалістів значно підвищує середнє значення.
Отже, медіана є більш підходящим показником типового доходу Python-розробника, ніж середнє.

Завдання 9. Аналіз популярності Python по віковим категоріям 
Мета: визначити, який відсоток респондентів у кожній віковій категорії
програмує на Python.

```python
# Перевірка колонок
cols = ["Age", "LanguageHaveWorkedWith"]
print(all(col in public.columns for col in cols))

public[cols].head()
```

```python
# Створюємо маску Python
python_mask = public["LanguageHaveWorkedWith"].str.contains("Python", na=False)
```

```python
#Групування по вікових категоріях
age_python = (
    public
    .assign(PythonUser=python_mask)
    .groupby("Age")["PythonUser"]
    .mean()
    .reset_index()
)

age_python["PythonPercent"] = age_python["PythonUser"] * 100

age_python = age_python[["Age", "PythonPercent"]]

age_python
```

Висновок
Найвищий відсоток Python-розробників спостерігається у віковій групі 18–24 роки (40%).
Далі показники поступово знижуються з віком.
Найменша частка Python-користувачів — серед респондентів 65+ років (~31.6%).
Отже, Python є найпопулярнішим серед молодших спеціалістів, але зберігає високу частку і в старших вікових групах, що свідчить про його універсальність та актуальність на всіх етапах кар’єри.

Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників
Мета: визначити, в яких індустріях найчастіше працюють респонденти,
які входять до 75-го перцентиля за компенсацією та працюють віддалено.

```python
#75-перцентль компенсації. Беремо тільки не порожні компенсації
comp = public["ConvertedCompYearly"].dropna()

p75 = comp.quantile(0.75)

print("75 перцентиль компенсації:", p75)
```

```python
# Фільтр: висока компенсація + remote
high_paid_remote = public[
    (public["ConvertedCompYearly"] >= p75) &
    (public["RemoteWork"] == "Remote")
]

print("Кількість високооплачуваних remote:", high_paid_remote.shape[0])
```

```python
# Визначаємо індустрії.У Stack Overflow колонка Industry
print("Industry" in public.columns)

high_paid_remote = high_paid_remote.dropna(subset=["Industry"])

industry_counts = (
    high_paid_remote["Industry"]
    .value_counts()
    .reset_index()
)

industry_counts.columns = ["Industry", "Count"]

industry_counts
```

Висновок
Серед високооплачуваних віддалених працівників (респонденти з компенсацією вище 75-го перцентиля та форматом роботи Remote) найпоширенішими індустріями є:
Software Development — 1186, 
Fintech — 190, 
Healthcare — 188, 
Other — 176, 
Internet, Telecom or Information Services — 138.
Це свідчить про те, що найбільші можливості для високооплачуваної віддаленої роботи зосереджені в ІТ-та фінансово-технологічних галузях, а також у суміжних цифрових секторах.
